- Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). 
- Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.


```python
# Elegeix el conversor
conversor = input("Escriu f per a convertir graus Farenheit o c per al convertir graus Celcius:")

# Conversor graus Farenheit a Celsius  i Kelvin
# La fórmula per passar de graus Farenheit a Celsius és: °C = (°F - 32) × (5/9)
if conversor == "f":
    F = int(input("Introdueix la temperatura en graus Farenheit"))
    C = round(5/9 * (F-32))
    K = C + 273.15
    print("La temperatura en graus Celsius és:")
    print(C)    
    print("La temperatura en graus Kelvin és:")
    print(K)


# Conversor graus Celsius a Farenheit i Kelvin
# La fórmula per passar de graus Celsius a Farenheit és: °F = °C × (9/5) + 32.
elif conversor == "c":
    C = int(input("Introdueix la temperatura en graus Celsius"))
    F = round(C * (9/5) + 32, 2)
    K = C + 273.15
    print("La temperatura en graus Farenheit és:")
    print(F)
    print("La temperatura en graus Kelvin és:")
    print(K)

else:
    print("No has introduit una opció vàlida")



```

### <span style="color: rgb(230, 0, 126);">Exercici 1: Diccionari invers amb duplicats</span>
Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/image007.png"/>


```python
# Aquest és el diccionari per a testar sense duplicats 
# dictionay_plus = {'x': 'apple', 'y': 'banana', 'z': 'grape'}

# Aquest és el diccionari per a testar amb duplicats
dictionay_plus = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
print("Aquest és el diccionari original:")
print(dictionay_plus)


# Aquest codi calcula llargades per a saber si hi ha duplicats:
# print("Aquest seria el diccionari invertit, eliminant els duplicats:")
reverse_dictionay_plus = dict((v, k) for k, v in dictionay_plus.items())
# print (reverse_dictionay_plus)

# print("Aquestos són els valors del diccionari original:")
# print(dictionay_plus.values())
# print(len(dictionay_plus.values()))

# print("Aquestos són els valors del diccionari invertit:")
# print(reverse_dictionay_plus.values())
# print(len(reverse_dictionay_plus.values()))

# Quan NO hi han duplicats, aquest codi crea un diccionari invertit 
if len(reverse_dictionay_plus.values()) == len(dictionay_plus.values()):
    print("El diccionari no té paraules repetides, s'ha invertit correctament")
    dictionay_plus = reverse_dictionay_plus


# Açò és el que fem quan hi han duplicats:
else:
    print("El diccionari té paraules repetides, es crearan llistes amb les paraules repetides")
    reverse_dictionay_plus_amb_duplicats = {}
    for key, value in dictionay_plus.items(): 

# Si el valor no existeix, el creem invertint clau i valor
        if  value not in reverse_dictionay_plus_amb_duplicats:
            reverse_dictionay_plus_amb_duplicats[value] = key

# Si existeix, hem de mirar si és una llista o un valor únic
        else:
# Si el valor ja és una llista, afegim la clau a la llista
            if type(value) == list:
                reverse_dictionay_plus_amb_duplicats[value].append(key)
# Si el valor no és una llista, el convertim en llista i afegim la clau anterior i la nova
            else:
                reverse_dictionay_plus_amb_duplicats[value] = [reverse_dictionay_plus_amb_duplicats[value], key]

    dictionay_plus = reverse_dictionay_plus_amb_duplicats


# Imprimim el diccionari que ens ha quedat
print("Aquest és diccionari que ens ha quedat")
print(dictionay_plus)



```

### <span style="color: rgb(230, 0, 126);">Exercici 2: Conversió de tipus de dades</span>
El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/image004.png"/>


```python
llista_per_convertir = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]
print("Hem de convertir la següent llista:")
print(llista_per_convertir)
print("Aquesta llista és de tipus:")
print(type(llista_per_convertir))

print("Ara intentem convertir els elements a decimals:")

llista_floats = []
llista_no_floats = []

for element in llista_per_convertir:
    if type(element) == tuple or type(element) == list:
        for subelement in element:
            try:
                llista_floats.append(float(subelement))
            except:
                llista_no_floats.append(subelement)
    else:
        try:
            llista_floats.append(float(element))
        except:
            llista_no_floats.append(element)

print("Els elements convertits a nombre decimal estan en la nova llista 'lista_floats' següent:")
print(llista_floats)

print("Els elements no convertits a nombre decimal estan en la nova llista 'lista_no_floats' següent:")
print(llista_no_floats)




```

### <span style="color: rgb(230, 0, 126);">Exercici 1: Comptador i endreçador de paraules d'un text</span>

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:




<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/imagen1data.png"/>


```python
## Comptador de paraules

# Demanem un text a l'usuari
text = input("Introdueix un text:")

# Eliminem els caràcters especials i passem el text a minúscules 
text = text.replace(".", "")
text = text.replace("(", "")
text = text.replace(")", "")
text = text.replace("¡", "")
text = text.replace("!", "")
text = text.replace(",", "")
text = text.replace(".", "")
text = text.replace(":", "")
text = text.replace(";", "")
text = text.lower()

# Separem creant una llista de les paraules, l'ordenem alfabèticmanent i la convertim en diccionari, per a eliminar els duplicats 
llista_paraules = text.split() 
llista_paraules = sorted(llista_paraules)
diccionari_contador_paraules = dict.fromkeys(llista_paraules)


# Contem les paraules i les afegim al nostre diccionari
for paraula in diccionari_contador_paraules:
    diccionari_contador_paraules.update({paraula: llista_paraules.count(paraula)})



## Diccionaris imbricats
# Creem una tupla amb l'abecedari
tupla_abecedari = ("a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m", "n", "o", "p", "q", "r", "s", "t", "u", "v", "w", "x", "y", "z")  

# I usem la tupla per a crear un primer diccionari de l'abecedari
diccionari_abecedari = dict.fromkeys(tupla_abecedari)


# Ara imbriquem el diccionari anterior dins el nostre diccionari de paraules
for lletra in tupla_abecedari:
    diccionari_abecedari.update({lletra: {}})
    for paraula in diccionari_contador_paraules:
        if paraula.startswith(lletra):
            diccionari_abecedari[lletra].update({paraula: diccionari_contador_paraules[paraula]})

print(diccionari_abecedari)



```
