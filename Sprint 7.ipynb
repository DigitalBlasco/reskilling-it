### <span style="color: rgb(230, 0, 126);"> Exercici 1: Calculadora de l'índex de massa corporal</span>

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:
>  [Índice de masa corporal IMC que es y como se calcula](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/).
- La funció ha de classificar el resultat en les seves respectives categories


```python

# Demanem les dades a l'usuari
altura = float(input("Introdueix la teva altura en cm"))

# Calculem l'IMC i el presentem a l'usuari
pes = float(input("Introdueix el teu pes en kg"))
IMC = pes / (altura/100)**2
print("El teu IMC es:")
print(IMC)

# Donem una interpretació de l'IMC
if IMC < 18.5:
    print("Tens un pes baix")
elif IMC < 24.9:
    print("Tens un pes normal")
elif IMC < 29.9:
    print("Tens sobrepés")
else:
    print("Tens obesitat")
```

- Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). 
- Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.


```python
# Elegeix el conversor
conversor = input("Escriu C per a convertir Farenheit a Celcius o escriu F per al conversor de Celcius a Farenheit:")

# Conversor graus Farenheit a Celsius 
# La fórmula per passar de graus Farenheit a Celsius és: °C = (°F - 32) × (5/9)
if conversor == "C":
    F = int(input("Introdueix la temperatura en graus Farenheit"))
    C = 5/9 * (F-32)
    print("La temperatura en graus Celsius és:")
    print(C)    

# Conversor graus Celsius a Farenheit
# La fórmula per passar de graus Celsius a Farenheit és: °F = °C × (9/5) + 32.
if conversor == "F":
    C = int(input("Introdueix la temperatura en graus Celsius"))
    F = C * (9/5) + 32
    print("La temperatura en graus Farenheit és:")
    print(F)

else:
    print("No has introduit una opció vàlida")




```

### <span style="color: rgb(230, 0, 126);">Exercici 3: Comptador de paraules d'un text</span>
- Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.


```python
# Comptador de paraules

# Demanem un text a l'usuari
text = input("Introdueix un text:")

# Eliminem els caràcters especials
text = text.replace(".", "")
text = text.replace("()", "")
text = text.replace(")", "")
text = text.replace("¡", "")
text = text.replace("!", "")
text = text.replace(",", "")
text = text.replace(".", "")
text = text.replace(":", "")
text = text.replace(";", "")


# Passem el text a minúscules
text = text.lower()

# Separem el text en paraules i en creem una llista
paraules = text.split() 

# Creem un diccionari amb les paraules de la llista
paraules = dict.fromkeys(paraules)

# Contem les paraules i les afegim al nostre diccionari
for paraula in paraules:
    paraules.update({paraula: text.count(paraula)})

# Mostrem el diccionari amb les paraules i el seu nombre d'aparicions
print (paraules)



```
