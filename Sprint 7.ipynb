### <span style="color: rgb(230, 0, 126);"> Exercici 1: Calculadora de l'índex de massa corporal</span>

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:
>  [Índice de masa corporal IMC que es y como se calcula](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/).
- La funció ha de classificar el resultat en les seves respectives categories


```python
def calcula_IMC():
    try:
        altura = float(input("Introdueix la teva altura en cm: "))
        pes = float(input("Introdueix el teu pes en kg: "))
    except ValueError:
        print("Si us plau, introdueix només números.")
        return

    IMC = round(pes / (altura / 100) ** 2, 2)
    print(f"\nEl teu IMC és: {IMC}")

    if IMC < 18.5:
        categoria = "Tens un pes baix"
    elif IMC < 24.9:
        categoria = "Tens un pes normal"
    elif IMC < 29.9:
        categoria = "Tens sobrepès"
    else:
        categoria = "Tens obesitat"

    print(categoria)
    return IMC, categoria

```

- Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). 
- Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.


```python
def convertir_temperatura():
    conversor = input("Escriu 'f' per convertir de Farenheit o 'c' per convertir de Celsius: ").lower()

    if conversor == "f":
        try:
            F = float(input("Introdueix la temperatura en graus Farenheit: "))
            C = round((F - 32) * 5 / 9, 2)
            K = round(C + 273.15, 2)
            print(f"La temperatura en graus Celsius és: {C}°C")
            print(f"La temperatura en graus Kelvin és: {K}K")
            return C, K
        except ValueError:
            print("Si us plau, introdueix un número vàlid.")
    
    elif conversor == "c":
        try:
            C = float(input("Introdueix la temperatura en graus Celsius: "))
            F = round(C * 9 / 5 + 32, 2)
            K = round(C + 273.15, 2)
            print(f"La temperatura en graus Farenheit és: {F}°F")
            print(f"La temperatura en graus Kelvin és: {K}K")
            return F, K
        except ValueError:
            print("Si us plau, introdueix un número vàlid.")
    
    else:
        print("No has introduït una opció vàlida ('f' o 'c').")

```

Per a utilitzar-la només cal cridar: convertir_temperatura()

```python
convertir_temperatura()
```

### <span style="color: rgb(230, 0, 126);">Exercici 3: Comptador de paraules d'un text</span>
- Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.


```python
from collections import Counter

def comptar_paraules():
    text = input("Introdueix un text: ")

    # Eliminem caràcters especials i convertim a minúscules
    for char in ".,()¡!,:;":
        text = text.replace(char, "")
    text = text.lower()

    # Separem les paraules i usem Counter per comptar-les
    llista_paraules = text.split()
    contador = Counter(llista_paraules)

    print(contador)
    return contador
```

### <span style="color: rgb(230, 0, 126);">Exercici 4: Diccionari invers</span>
- Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. 
- Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.




<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/sprint7.n1.e4.png"/>


```python

def invertir_diccionari(diccionari_original):
    valors = list(diccionari_original.values())

    # Comprovem si hi ha valors duplicats
    if len(valors) != len(set(valors)):
        print("No es pot invertir el diccionari perquè hi ha valors duplicats.")
        return None

    # Invertim claus i valors
    diccionari_invertit = {valor: clau for clau, valor in diccionari_original.items()}
    print("Diccionari invertit correctament:")
    return diccionari_invertit
```

```python
# Aquest és el diccionari per a testar amb duplicats
dictionay_plus = {'x': 'apple', 'y': 'banana', 'z': 'strawberry'}
print("Aquest és el diccionari original:")
print(dictionay_plus)


# Aquest codi calcula llargades per a saber si hi ha duplicats:
reverse_dictionay_plus = invertir_diccionari(dictionay_plus)
print(reverse_dictionay_plus)
```

### <span style="color: rgb(230, 0, 126);">Exercici 1: Diccionari invers amb duplicats</span>
Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/image007.png"/>


```python


def invertir_diccionari(diccionari_original):
    valors = list(diccionari_original.values())

    # Comprovem si hi ha valors duplicats
    if len(valors) == len(set(valors)):
        # No hi ha valors duplicats → podem invertir directament
        diccionari_invertit = {valor: clau for clau, valor in diccionari_original.items()}
        print("Diccionari invertit correctament (sense valors duplicats):")
    else:
        # Hi ha valors duplicats → agrupem claus en llistes
        print("El diccionari té valors duplicats. S'agrupen les claus en llistes.")
        diccionari_invertit = {}
        for clau, valor in diccionari_original.items():
            if valor not in diccionari_invertit:
                diccionari_invertit[valor] = clau
            else:
                if isinstance(diccionari_invertit[valor], list):
                    diccionari_invertit[valor].append(clau)
                else:
                    diccionari_invertit[valor] = [diccionari_invertit[valor], clau]

    return diccionari_invertit


```

```python
# Aquest és el diccionari per a testar amb duplicats
dictionay_plus = {'x': 'apple', 'y': 'banana', 'z': 'strawberry', 'a': 'kiwi', 'b': 'banana', 'c': 'kiwi', 'd': 'kiwi', 'e': 'kiwi', 'f': 'kiwi', 'g': 'kiwi', 'h': 'kiwi', 'i': 'kiwi', 'j': 'kiwi', 'k': 'kiwi', 'l': 'kiwi', 'm': 'kiwi', 'n': 'kiwi', 'o': 'kiwi', 'p': 'kiwi', 'q': 'kiwi', 'r': 'kiwi',}
print("Aquest és el diccionari original:")
print(dictionay_plus)


# Aquest codi calcula llargades per a saber si hi ha duplicats:
print("Aquest seria el diccionari invertit, eliminant els duplicats:")
reverse_dictionay_plus = invertir_diccionari(dictionay_plus)
print (reverse_dictionay_plus)
```

### <span style="color: rgb(230, 0, 126);">Exercici 2: Conversió de tipus de dades</span>
El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/image004.png"/>


```python
def separar_convertibles_a_float(llista_entrada):
    llista_floats = []
    llista_no_floats = []

    for element in llista_entrada:
        if isinstance(element, (tuple, list)):
            for subelement in element:
                try:
                    llista_floats.append(float(subelement))
                except:
                    llista_no_floats.append(subelement)
        else:
            try:
                llista_floats.append(float(element))
            except:
                llista_no_floats.append(element)

    return llista_floats, llista_no_floats

```

```python
llista_per_convertir = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

separar_convertibles_a_float(llista_per_convertir)






```

### <span style="color: rgb(230, 0, 126);">Exercici 1: Comptador i endreçador de paraules d'un text</span>

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:




<img src="https://lablascovegmenu.wordpress.com/wp-content/uploads/2025/02/imagen1data.png"/>


```python
# Ara construïm la funció que genera el diccionari imbricat
def diccionari_imbricat_per_lletra():
    diccionari_contador_paraules = comptar_paraules()

    diccionari_abecedari = {}

    for paraula, freq in diccionari_contador_paraules.items():
        lletra_inicial = paraula[0]
        if lletra_inicial not in diccionari_abecedari:
            diccionari_abecedari[lletra_inicial] = {}
        diccionari_abecedari[lletra_inicial][paraula] = freq

    return diccionari_abecedari


# Cridem la funció i imprimim el resultat
resultat = diccionari_imbricat_per_lletra()

from pprint import pprint
pprint(resultat, sort_dicts=True)
```
